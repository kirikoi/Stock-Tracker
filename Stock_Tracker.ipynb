```python
import yfinance as yf
import pandas as pd
```

```python
def get_all_data():
    tickers = ['MSFT', 'AAPL', 'TSLA','GOOG', 'WMT', 'NVDA', 'V', 'NKE', 'VZ', 'AXP', 'IBM','META','AVGO','JPM','SHEL','BP']
    fields = ['symbol', 'longName', 'currency', 'open', 'dayLow', 'dayHigh',
          'previousClose', 'volume', 'averageVolume10days', 'marketCap','fiftyTwoWeekLow','fiftyTwoWeekHigh','52WeekChange']
    rows = []
    for symbol in tickers:
        try:
            info = yf.Ticker(symbol).info
        except Exception as e:
            print(symbol, "failed:", e)
            continue
        row = {}
        for f in fields:
            row[f] = info.get(f)
        row['diff_from_52w_low'] = info.get('previousClose') - info.get('fiftyTwoWeekLow')
        rows.append(row)
    return pd.DataFrame(rows)
```
